```python
new_loc = 'C:\\Users\\Pratt\\Desktop\\HKUST-RA\\Database Construction P2\\current.csv'
```

```python
current_df = pd.read_csv(new_loc)
```

```python
unique_current_df  = current_df.drop_duplicates(subset='gvkey')
unique_current_df.reset_index(drop=True, inplace=True)
```

```python
unique_current_df
```

```python
unique_df
```

```python
merged = unique_df.merge(unique_current_df, on='gvkey', how='inner')
```

```python
merged
```

```python
import json

json_data = merged[['companyname', 'weburl']].to_dict(orient='records')

output_path = 'C:\\Users\\Pratt\\Desktop\\HKUST-RA\\Database Construction P2\\company_weburl.json'
with open(output_path, 'w', encoding='utf-8') as f:
    json.dump(json_data, f, ensure_ascii=False, indent=2)

print(f"JSON 文件已保存到: {output_path}")
print(f"共 {len(json_data)} 条记录")
```
